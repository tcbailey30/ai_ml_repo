```python
# Create a datatype summary table
datatype_summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes,
    "Missing Values": df.isnull().sum().values,
    "Unique Values": df.nunique().values
})

datatype_summary
```

```python
datatype_summary.sort_values(
    by="Unique Values", ascending=True
)

datatype_summary.to_csv("/Users/tcbailey/Documents/ai_ml_repo/supervised_learning/projects/airbnb_prices_model/data/processed/datatype_summary.csv", index=False)
```
